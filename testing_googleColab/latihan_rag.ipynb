```python
from pypdf import PdfReader
import requests
import json

# 1. Mula membaca fail PDF (Sama seperti sebelum ini)
print("Sedang mengekstrak teks daripada dokumen_sdg.pdf...")
reader = PdfReader("dokumen_sdg.pdf")
teks_pdf = ""
for page in reader.pages:
    teks_pdf += page.extract_text()

# Hadkan teks PDF untuk keselamatan memori komputer lokal
teks_pdf = teks_pdf[:2000] 

print("Fail PDF sedia untuk disoal!\n")

# 2. FUNGSI DINAMIK: Mengambil soalan terus daripada anda semasa kod dijalankan
soalan_anda = input("Masukkan soalan anda tentang dokumen ini: ")

print(f"\nSedang memproses soalan: '{soalan_anda}' bersama model Qwen...")

# Rangka Prompt RAG dengan soalan dinamik anda
prompt_rag = f"""
Anda adalah AI pembantu kelestarian pasukan hackathon. Jawab soalan pengguna berdasarkan KONTEKS dokumen yang diberikan sahaja. 
Jika jawapan tiada dalam konteks dokumen, katakan anda tidak tahu.

KONTEKS DOKUMEN:
{teks_pdf}

SOALAN: {soalan_anda}
JAWAPAN:
"""

# 3. Hantar prompt RAG ke Ollama lokal
url = "http://localhost:11434/api/generate"
payload = {
    "model": "qwen",  
    "prompt": prompt_rag,
    "stream": False
}

try:
    response = requests.post(url, json=payload)
    result = json.loads(response.text)
    
    if "error" in result:
        print("\n--- Ralat dari Sistem Ollama ---")
        print(result["error"])
    else:
        print("\n--- Hasil Jawapan RAG Dinamik ---")
        print(result["response"])
        
except Exception as e:
    print("\nGagal menyambung ke Ollama. Info ralat teknikal:", e)
```
