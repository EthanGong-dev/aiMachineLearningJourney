```python
import requests
import json

url = "http://localhost:11434/api/generate"

# Nota: Pastikan nama model ini sama tepat dengan nama model yang anda download dalam Ollama
payload = {
    "model": "qwen3:1.7b",  
    "prompt": "Sebut 'Helo Tim Hackathon! Sistem AI kita sudah sedia.'",
    "stream": False
}

try:
    response = requests.post(url, json=payload)
    result = json.loads(response.text)
    
    # Jika ada error dari Ollama, kita paparkan di sini
    if "error" in result:
        print("Ralat dari Ollama:", result["error"])
    else:
        print("Jawapan dari Qwen:", result["response"])
        
except Exception as e:
    print("Gagal menyambung ke Ollama. Pastikan aplikasi Ollama sudah dibuka!")
    print("Info ralat:", e)
```
